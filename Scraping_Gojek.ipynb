```python
import pandas as pd
import numpy as np
from google_play_scraper import Sort, reviews
import os

# Mengambil data review dari aplikasi Gojek
result, continuation_token = reviews(
    'com.gojek.app',
    lang='id',
    country='id',
    sort=Sort.MOST_RELEVANT,
    count=7000,
)

# Membuat DataFrame
df = pd.DataFrame(np.array(result), columns=['review'])
df = df.join(pd.DataFrame(df.pop('review').tolist()))

# Memilih kolom yang ingin disimpan
data = df[['reviewId', 'userName', 'score', 'content']]  # 'content' berisi teks review

file_path = '/content/drive/My Drive/data_scraping_gojek.xlsx'

# Menyimpan DataFrame ke file Excel di Google Drive
data.to_excel(file_path, index=False)

print(f"File berhasil disimpan di {file_path}")
```

```python
import pandas as pd
import re

# Membaca file Excel dari Google Drive
file_path = '/content/drive/My Drive/data_scraping_gojek.xlsx'
data = pd.read_excel(file_path)

# Fungsi cleansing untuk setiap review
def cleansing_text(text):
    # Mengubah teks menjadi lowercase
    text = text.lower()
    # Menghapus karakter khusus dan tanda baca
    text = re.sub(r'[^\w\s]', '', text)
    # Menghapus whitespace yang berlebihan
    text = re.sub(r'\s+', ' ', text).strip()
    return text

# Melakukan cleansing pada kolom 'content'
data['cleaned_content'] = data['content'].apply(cleansing_text)

# Menghapus baris yang mungkin kosong setelah cleansing
data = data.dropna(subset=['cleaned_content'])

# Menyimpan hasil cleansing kembali ke file Excel
output_path = '/content/drive/My Drive/cleansed_data_scraping_gojek.xlsx'
data.to_excel(output_path, index=False)

print(f"File hasil cleansing berhasil disimpan di {output_path}")
```

```python
import pandas as pd
import re

# Membaca file Excel yang sudah dibersihkan dari Google Drive
file_path = '/content/drive/My Drive/cleansed_data_scraping_gojek.xlsx'
data = pd.read_excel(file_path)

# Fungsi tokenisasi untuk setiap review menggunakan regex
def tokenize_text(text):
    # Tokenisasi dengan memisahkan kata menggunakan regex
    tokens = re.findall(r'\b\w+\b', text)
    return tokens

# Melakukan tokenisasi pada kolom 'cleaned_content' dan menyimpan hasilnya dalam variabel
tokenized_reviews = data['cleaned_content'].apply(tokenize_text).tolist()

# Menampilkan hasil tokenisasi
for i, tokens in enumerate(tokenized_reviews[:10]):  # Cetak 10 review pertama sebagai contoh
    print(f"Review {i+1} Tokens: {tokens}")
```

```python
import pandas as pd
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory

# Inisialisasi stemmer Sastrawi
factory = StemmerFactory()
stemmer = factory.create_stemmer()

# Path ke dataset
file_path = '/content/drive/My Drive/categorized_data_gojek.xlsx'

# Baca dataset
df = pd.read_excel(file_path)

# Pastikan dataset memiliki kolom yang sesuai
text_column = 'cleaned_content'  # Nama kolom yang akan di-stemming
if text_column not in df.columns:
    raise ValueError(f"Kolom '{text_column}' tidak ditemukan dalam dataset.")

# Fungsi stemming
def apply_stemming(text):
    if isinstance(text, str):  # Pastikan input adalah string
        return stemmer.stem(text)
    return text

# Terapkan stemming ke kolom
df[f'{text_column}_stemmed'] = df[text_column].apply(apply_stemming)

# Simpan hasil ke file baru
output_file = '/content/drive/My Drive/categorized_data_gojek_stemmed.xlsx'
df.to_excel(output_file, index=False)

print(f"Dataset dengan stemming selesai disimpan di {output_file}")
```

```python
import pandas as pd
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

# Download stopwords NLTK jika belum ada
nltk.download('stopwords')
nltk.download('punkt')
# Download the punkt_tab data
nltk.download('punkt_tab') # This line was added to download the missing data package.

# Path ke dataset
file_path = '/content/drive/My Drive/categorized_data_gojek_normalized.xlsx'

# Baca dataset
df = pd.read_excel(file_path)

# Pastikan dataset memiliki kolom teks
text_column = 'cleaned_content_stemmed_normalized'  # Ganti dengan nama kolom teks pada dataset Anda

if text_column not in df.columns:
    raise ValueError(f"Kolom '{text_column}' tidak ditemukan dalam dataset.")

# Inisialisasi stopwords
stop_words = set(stopwords.words('indonesian'))  # Ganti 'indonesian' dengan bahasa lain jika diperlukan

# Fungsi untuk menghapus stopwords
def remove_stopwords(text):
    if isinstance(text, str):  # Pastikan input adalah string
        words = word_tokenize(text)
        filtered_words = [word for word in words if word.lower() not in stop_words]
        return ' '.join(filtered_words)
    return text

# Terapkan fungsi ke kolom teks
df[f'{text_column}_stopwords'] = df[text_column].apply(remove_stopwords)

# Simpan dataset hasil proses
output_file = '/content/drive/My Drive/categorized_data_stopwords.xlsx'
df.to_excel(output_file, index=False)

print(f"Dataset tanpa stopwords telah disimpan di {output_file}")
```
